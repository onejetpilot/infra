## Схема данных Olist

```text
customers 1 ── N orders 1 ── N order_items N ── 1 products
                       │              │
                       │              └── N:1 sellers
                       ├── 1:N order_payments
                       └── 1:N order_reviews
products N ── 1 product_category_translation
customers/sellers N ── 1 geolocation (по zip_code_prefix после агрегации)
```

Рабочие типизированные объекты находятся в `staging`. `order_id` — ключ заказа; `(order_id, order_item_id)` — grain позиции; у платежей grain `(order_id, payment_sequence)`. Перед JOIN всегда проверяйте фактическую уникальность, а не полагайтесь только на рисунок.
